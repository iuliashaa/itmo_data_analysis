```python
#  Активируем библиотеки
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
#  Указываем колонку numb.obs как index_col
df = pd.read_csv('/content/drive/MyDrive/ITMO/data analysis/lab_5/beverage_r.csv', sep = ';', index_col='numb.obs')
```

```python
# # Задача 1
# Разделить таблицу данных так, чтобы в обучающую выборку попало 66% наблюдений, а в тестовую выборку 33% наблюдений

# Решение

df_train, df_test = train_test_split(df, test_size=0.33, random_state=42)

print(f"Размер обучающей выборки (66%): {len(df_train)} наблюдений")
print(f"Размер тестовой выборки (33%): {len(df_test)} наблюдений")
```

```python
# Задача 2
# Разделить таблицу данных на четыре одинаковых части.

# Решение

# Разделяем df на две части по 50%
df_half1, df_half2 = train_test_split(df, test_size=0.5, random_state=42)

# Разделяем первую половину на две части по 25% от оригинала
df_part1, df_part2 = train_test_split(df_half1, test_size=0.5, random_state=42)

# Разделяем вторую половину на две части по 25% от оригинала
df_part3, df_part4 = train_test_split(df_half2, test_size=0.5, random_state=42)

print(f"Размер первой части: {len(df_part1)} наблюдений")
print(f"Размер второй части: {len(df_part2)} наблюдений")
print(f"Размер третьей части: {len(df_part3)} наблюдений")
print(f"Размер четвертой части: {len(df_part4)} наблюдений")
```

```python
# Задача 3
# Разделить таблицу данных на две стратифицированные подвыборки.
# По переменным кока-кола и минералка.

# Решение

# Пусть минеральной водой будет напиток SPRITE

# Стратификация по переменным COKE и SPRITE
stratify_cols = df[['COKE', 'SPRITE']].apply(lambda x: f"{x['COKE']}_{x['SPRITE']}", axis=1)

# Создаем объединенную переменную для стратификации, чтобы посмотреть ее распределение
df['COKE_SPRITE_STRAT'] = df['COKE'].astype(str) + '_' + df['SPRITE'].astype(str)

print("Распределение категорий в объединенной переменной 'COKE_SPRITE_STRAT':")
print(df['COKE_SPRITE_STRAT'].value_counts())

# Разделение данных на две стратифицированные подвыборки (50%/50%)
df_1, df_2 = train_test_split(df, test_size=0.5, stratify=stratify_cols, random_state=42)

print(f"\nРазмер первой подвыборки: {len(df_1)}")
print(f"Размер второй подвыборки: {len(df_2)}")
print("\nРаспределение в первой подвыборке:")
print(df_1[['COKE', 'SPRITE']].value_counts().sort_index())
print("\nРаспределение во второй подвыборке:")
print(df_2[['COKE', 'SPRITE']].value_counts().sort_index())
```
