```python
# Задача 1 (2.25) Найти вероятность получить 12 очков хотя бы один раз при n бросаниях двух игральных костей.

# Решение

import random

# Устанавливаем зерно для воспроизводимости результатов
random.seed(42)

# Параметры эксперимента
n = 50  # количество бросков в одном эксперименте
n_exp = 100000  # количество экспериментов

# Счетчик успешных экспериментов
k_success = 0

# Проводим n_exp экспериментов
for i in range(n_exp):
    success_in_experiment = False

    # В каждом эксперименте делаем n бросков двух костей
    for j in range(n):
        # Бросаем две кости
        dice1 = random.randint(1, 6)
        dice2 = random.randint(1, 6)

        # Проверяем, равна ли сумма 12
        if dice1 + dice2 == 12:
            success_in_experiment = True
            break  # если нашли 12, прерываем цикл

    # Если в эксперименте была хотя бы одна 12, увеличиваем счетчик
    if success_in_experiment:
        k_success += 1

# Вычисляем вероятность
probability = k_success / n_exp
print(f'Вероятность получить 12 очков хотя бы раз за {n} бросков: {probability:.3f}')

# Сравнение с аналитическим решением
analytical = 1 - (35/36)**n
print(f'Аналитическое решение: {analytical:.3f}')
print(f'Разница: {abs(probability - analytical):.3f}')
```

```python
# Задача 2 (3.9) Стержень длины l наудачу разламывается на две части, после
# чего большая из частей опять разламывается надвое в наудачу выбранной точке.
# Найти вероятность того, что из полученных частей можно составить треугольник.

# Решение

import random
import math

# Устанавливаем зерно для воспроизводимости результатов
random.seed(42)

# Параметры эксперимента
l = 1.0  # длина стержня
n_exp = 100000  # количество экспериментов

# Счетчик успешных экспериментов
k_success = 0

# Проводим n_exp экспериментов
for i in range(n_exp):
    # Первый разлом - случайная точка на отрезке [0, l]
    break_point1 = random.uniform(0, l)

    # Две части после первого разлома
    part1 = break_point1
    part2 = l - break_point1

    # Находим большую часть
    if part1 > part2:
        long_part = part1
    else:
        long_part = part2

    # Второй разлом - случайная точка на большей части
    break_point2 = random.uniform(0, long_part)

    # Получаем три части:
    # - меньшая часть от первого разлома
    # - две части от второго разлома большего отрезка
    if part1 > part2:
        # part1 дробится на две части
        a = break_point2          # первая часть от part1
        b = part1 - break_point2  # вторая часть от part1
        c = part2                 # оставшаяся часть от первого разлома
    else:
        # part2 дробится на две части
        a = part1                 # оставшаяся часть от первого разлома
        b = break_point2          # первая часть от part2
        c = part2 - break_point2  # вторая часть от part2

    # Проверяем условие существования треугольника:
    # сумма любых двух сторон должна быть больше третьей
    if (a + b > c) and (a + c > b) and (b + c > a):
        k_success += 1

# Вычисляем вероятность методом Монте-Карло
probability_monte_carlo = k_success / n_exp

# Правильный аналитический ответ
probability_analytical = math.log(4) - 1  # ln(4) - 1

print('РЕЗУЛЬТАТЫ:')
print(f'Метод Монте-Карло: {probability_monte_carlo:.4f}')
print(f'Аналитическое решение: {probability_analytical:.4f}')
print(f'Разница: {abs(probability_monte_carlo - probability_analytical):.4f}')
```
